```python
rating = df["Customer_Rating"].mean()

if rating >= 4.5:
  print("excellent")

elif rating >= 4.0:
    print("Good")

elif rating >= 3.0:
   print("average")

else:
  print("needs improvement")
```

```python
#for loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
    print("Good Rating")
  else:
     print("needs improvement")
```
