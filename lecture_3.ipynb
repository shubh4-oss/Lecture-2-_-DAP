```python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Average revenue is high")

  if rating >= 4:
    print("customer stastifaction is also good")
  else:
    print("customer staisfaction needs improvement")
else:
  print("revenue is low")
```

```python
rating = df["Customer_Rating"].mean()

if rating >= 4.5:
  print("excellent")

elif rating >= 4.0:
    print("Good")

elif rating >= 3.0:
   print("average")

else:
  print("needs improvement")
```
